```python
#  NORMAL DISTRIBUTION — Bell Curve
data = np.random.normal(loc=170, scale=10, size=10000)
print(f"length is {len(data)}")
print(f"mean is {data.mean()}")
print(f"mean is {data.mean():.1f}")
print(f"std is {data.std():.1f}")
plt.figure(figsize=(10, 5))
plt.hist(data, bins=50,color="#339af0", edgecolor="white", density=False)
plt.axvline(170, color="red", linestyle="--", linewidth=2, label="Mean (170)")
plt.axvline(160, color="orange", linestyle=":", label="-1 Std (160)")
plt.axvline(180, color="orange", linestyle=":", label="+1 Std (180)")
plt.legend()
plt.title("Normal Distribution (Heights)")
plt.xlabel("Height (cm)")
plt.ylabel("Probability Density")
plt.show()
```

```python
#  GRADIENT DESCENT — MODEL SEEKHTA KAISE HAI

X = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([20, 35, 45, 55, 70, 80, 85, 95])


w = 0.0
# weight =

b = 0.0
# bias

lr = 0.01
# lr = LEARNING RATE

epochs = 100
# epochs

losses = []

print("Training started...\n")

for epoch in range(epochs):
    y_pred = w * X + b
    # prediction = weight × hours + bias

    # ── Step 2: ERROR (Loss)
    loss = np.mean((y_pred - y) ** 2)
    # MSE = Mean Squared Error
    #
    # Step by step:
    # (y_pred - y) =

    losses.append(loss)


    # ── Step 3: GRADIENT
    dw = np.mean(2 * (y_pred - y) * X)
    db = np.mean(2 * (y_pred - y))
    w = w - lr * dw
    b = b - lr * db

if epoch % 20 == 0:
    print(f"Epoch {epoch:3d}: w={w:.4f}, b={b:.4f}, Loss={loss:.2f}")


print(f"\n Training complete!")
print(f"Final: w={w:.4f}, b={b:.4f}")
print(f"Model: marks = {w:.2f} × hours + {b:.2f}")


```
